#### 3. Data quality checks


```python
#It's better for me to convert training and testing_data to dataframes to handle pandas operations
X = pd.DataFrame(data={
    'text':train_data['text'],
    'label':train_data['label']})
y = pd.DataFrame(data={
    'text':test_data['text'],
    'label':test_data['label']
})
```

```python
print('Train data 5 samples')
X.head()
```

```python
print('Test Data 5 samples')
y.head()
```

```python
### Missing value count
for column in X.columns:
    missing = X[column].isna().sum()
    print(f'Missing value count for column "{column}" in Training data is: {missing}')
```

```python
for column in y.columns:
    missing = y[column].isna().sum()
    print(f'Missing value count for column "{column}" in Testing Data is: {missing}')
```

```python
### Empty whitespaces only questions
col = X['text'].astype(str)
empty_or_ws = col.str.strip('').eq('')
count = empty_or_ws.sum()
print(f'Empty or whitespace-only rows in Train Data: {count}')
```

```python
### Empty whitespaces only questions
col = y['text'].astype(str)
empty_or_ws = col.str.strip('').eq('')
count = empty_or_ws.sum()
print(f'Empty or whitespace-only rows in Test Data: {count}')
```

```python
### Duplicated questions
print('Duplicated row count for Train data:')
print(X['text'].duplicated().sum())
```

```python
print('Duplicated row count for Test data:')
print(y['text'].duplicated().sum())
```

```python
label_counts = X.groupby('text')['label'].nunique()
label_counts
```

```python
X['label'].value_counts().index
```

```python
X['label'].value_counts().head()
```

```python
X['label'].value_counts().tail()
```

```python
### Vizualize distribution
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(X['label'], bins=50, edgecolor='black')
ax.set_xlabel('label')
ax.set_ylabel('count')
ax.set_title('Distribution of amount')
plt.tight_layout()
plt.show()
```

```python
### - Compare minimum, maximum and average examples per category.
counts = X['label'].value_counts()
print('min:',counts.min())
print('max:',counts.max())
```

```python
print('smallest class:', counts.idxmin(), '→', counts.min())
print('largest  class:', counts.idxmax(), '→', counts.max())
```
