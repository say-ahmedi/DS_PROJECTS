* **Task 1: Data Representation**
  * *Instructions:* Create a Python dictionary representing a mini-dataset of 5 credit card transactions. The keys should be `Transaction_ID`, `Merchant_Type` (e.g., Groceries, Electronics, Travel), and `Is_Fraud` (Yes/No).
  * *Acceptance Criteria:* The dictionary is successfully created and can be printed in a readable format.

```python
data = {'Transaction_ID':[0,1,2,3,4],'Merchant_Type':['Groceries','Groceries','Travel','Electronics','Travel'],'Is_Fraud':['Yes','Yes','No','No','No']}
for key,value in data.items():
    print(key,value)
print('Transaction_ID->Quantitative variable')
print('Merchant_Type->Categorical variable')
print('Is_Fraud->Categorical variable')
```

```python
for i in range(len(data['Transaction_ID'])):
    print('ID:',data['Transaction_ID'][i],
    'Merchant:',data['Merchant_Type'][i],
    'Fraud:',data['Is_Fraud'][i])
    
```

* **Task 2: Bar Chart Visualization**
  * *Instructions:* Count the frequency of each `Merchant_Type` in your dataset. Use `matplotlib.pyplot` to create a bar chart showing the count of each category.
  * *Acceptance Criteria:* A bar chart renders successfully with an X-axis (Merchant Types), a Y-axis (Counts), and a title.

```python
### Count The Frequency
freq_merchant_type = {}
for m_type in data['Merchant_Type']:
    freq_merchant_type[m_type] = freq_merchant_type.get(m_type,0) + 1

freq_merchant_type
```

```python
### Bar Chart Visualization
colors = plt.cm.tab10(range(len(freq_merchant_type)))

plt.bar(freq_merchant_type.keys(),freq_merchant_type.values(),color=colors,edgecolor='black')
plt.title('Merchant Type Frequency Bar Chart')
plt.xlabel('Merchant Types')
plt.ylabel('Counts')
plt.xticks(rotation=45) 
plt.show()
```

```python
### Task 1
dct = {'Name':['Jigga','Bigga','Digga','Ligga','Migga'],
       'Favorite_Fruit':['Banana','Apple','Banana','Apple','PineApple'],
       'Is_Student':['Yes','Yes','No','No','No']}



fruit_freq = dict()
for fruit in dct['Favorite_Fruit']:
    fruit_freq[fruit] = fruit_freq.get(fruit,0) + 1
    
colors = plt.cm.tab10(range(len(fruit_freq)))

plt.bar(fruit_freq.keys(),fruit_freq.values(),color=colors)
plt.title('Fruit Type Frequency Distribution')
plt.xlabel('Fruit Types')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.show()
```

* **Task 5: Universal (Set Logic / Venn Diagrams)**
  * *Instructions:* Create two lists: `math_students` and `science_students`. Use Python `set` operations (`intersection`, `union`, `difference`) to find: 1) Students taking both, 2) Students taking at least one, 3) Students taking *only* math.
  * *Acceptance Criteria:* The script successfully prints the three distinct lists using set operators (`&`, `|`, `-`).


```python
math_students = ['Johan','Bohan','Tohan','Mohan']
science_students = ['Johan','Bojan','Mohan','Nijjan']

both = list(set(math_students)&set(science_students))
print('Students taking both subjects:',both)

only_math = list(set(math_students).difference(set(science_students)))
print('Students who chose only math',only_math)
at_least_one_subject = list(set(math_students).union(set(science_students)))
print('Students with at least_one_subject:',at_least_one_subject)
```

* **Task 6: Capstone (Antifraud Application)**
  * *Instructions:* Create a dictionary representing 5 credit card transactions. Keys: `Transaction_ID`, `Merchant_Type`, `Is_Fraud`. Count the frequency of each `Merchant_Type` and plot a bar chart. Then, create two sets: `high_value_transactions` and `international_transactions`. Use set logic to find transactions that are BOTH high value AND international.
  * *Acceptance Criteria:* A bar chart renders successfully. The set logic correctly identifies the overlap.

```python
import random
```

```python

transactions = {'Transaction_ID':random.sample(range(10000,20000),5),
       'Merchant_Type':['Groceries','Accessories','Groceries','Gadgets','Commercial'],
       'Is_Fraud':['No','Yes','Yes','No','Yes'],
       'Amount':[800,1500,400,950,1400],
       'Location':['Local','International','International','Local','Local']
       }

freq_merchant_type = {}
for merch in transactions['Merchant_Type']:
    freq_merchant_type[merch] = freq_merchant_type.get(merch,0) + 1
freq_merchant_type
```

```python
colors = plt.cm.tab10(range(len(freq_merchant_type)))
plt.bar(freq_merchant_type.keys(),freq_merchant_type.values(),color=colors)
plt.xlabel('Merch Types')
plt.ylabel('Frequency')
plt.title('Merch Type Frequency Distribution')
plt.show()
```

```python
high_value_transactions = list(id for id,am in zip(transactions['Transaction_ID'],transactions['Amount']) if am>=1000)
international_transactions = list(id  for id,loc in zip(transactions['Transaction_ID'],transactions['Location']) if loc=='International')
```

```python
value = set(international_transactions) & set(high_value_transactions)
print('International and high value transaction',value)
```
