C answer is ->because a 1-d array has only one axis and .T does exactly one thing: reverse the order of the axes tuple.
a_1d.shape is (3,) → one axis.

Transpose reverses (3,) → (3,). There is nothing to permute.

ndim stays 1; no new dimension is inserted.

NumPy's 1‑D array is a rank‑1 tensor — conceptually neither a row nor a column vector

Does NumPy always reject arrays with different shapes?

You intend to subtract per-feature means (D,) from a batch (N, D), but accidentally use a shape like (N,). Broadcasting may align it against the wrong axis or fail confusingly, producing per-sample instead of per-feature normalization.

Shapes (N, 1) and (1, M) broadcast to (N, M). If N and M are large, this can explode memory or compute—e.g., accidentally computing all pairwise distances when you wanted elementwise differences.

A mask of shape (N,) applied to data (N, D) may broadcast along the wrong axis, masking features instead of samples—or vice versa.
