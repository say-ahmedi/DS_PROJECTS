```python
import numpy as np

v = np.array([3, -2, 5])
a = np.array([1, 2, 3])
b = np.array([3, 2, 1])
w = np.array([-4, 0, 3])

# Calculate L1 norm
print("L1 norm of v:",np.linalg.norm(v,ord=1))
print("L2 norm of v:",np.linalg.norm(v,ord=2))

# Calculate L2 norm
print('Difference vector:',a-b)
print('Euclidean Distance:',np.linalg.norm(a-b,ord=2))


# Verify your handwritten answers
print("L1 norm of w:",np.linalg.norm(w,ord=1))
print("L2 norm of w:",np.linalg.norm(w,ord=2))
```
