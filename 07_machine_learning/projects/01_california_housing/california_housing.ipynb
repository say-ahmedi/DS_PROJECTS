### Your objective
Predict the median house value of a California census block group using features such as median income, house age, and average number of rooms.

This is a regression problem because the target is continuous.

    Understanding the dataset and its variables.
    Performing exploratory data analysis (EDA).
    Identifying outliers and investigating feature distributions.
    Splitting data into training and testing sets.
    Training a linear regression model.
    Evaluating predictions using MAE, RMSE, and \(R^2\).
    Comparing ordinary linear regression with Ridge regression.
